### defaultdict

A defaultdict is like a regular dictionary, except that when you try to look up a key it doesn’t contain, it first adds a value for it using a zero-argument function you provided when you created it. In order to use defaultdicts, you have to import them from collections.

In high-performance AI systems—like a recommendation engine at Netflix or Amazon—you are constantly grouping and aggregating data. Using a standard dictionary with if key not in d: checks or .get() calls adds overhead to every single operation

1. The "Why": Production Efficiency at Scale
Why does Google or Meta use defaultdict (or similar structures) in their data pipelines?

Branch Prediction: Modern CPUs try to guess the path your code will take. Using if key not in d: introduces a branch. In a loop of 100 million tokens (common in LLM preprocessing), these branch misses add up to significant latency. defaultdict moves the "missing key" logic into the C-implemented backend of Python, making it faster and cleaner.

Cleaner Aggregations: When building an Inverted Index (the core of search engines), you need to map words to a list of document IDs. defaultdict(list) allows you to append directly without boilerplate.


TECHNICAL PART

A defaultdict takes a callable (a function) as its first argument. This is known as the default_factory.

defaultdict(int): The factory is int(), which returns 0.

defaultdict(list): The factory is list(), which returns an empty list [].

defaultdict(lambda: "Missing"): A custom factory that returns a specific string.

Crucial Logic Note: The factory is only called when you access a key using d[key]. If you use d.get(key), it will still return None (or your specified default), and it will not create the entry in the dictionary.

defaultdict factory (like lambda) must be nullary (no arguments). 
It doesn't auto-compute values; it only provides defaults on missing key access

EX: `square_of_num = defaultdict(lambda: 0)  # or any default`

```python
from collections import defaultdict

items: list[tuple[str,int]] = [('a',1),('b',2),('a',3),('c',4),('a',5),('b',6),]

# here you can clearly see that in the list a specific key like have multiple values so we want to convert it into dict of key and list of values which the same key acquire like 'a' = [2,3,5]

d:defaultdict[str,list] = defaultdict(list) # it means it initiates a empty list for every key when it is accessed also remember the friend logic where you manually using for loop adds the friends list.

for k,v in items:
    d[k].append(v)
   
print(d)

# if you want to print it as normal dictionary just cast it back to dict
print(dict(d))


```

### COUNTERS

You can pass any iterable to the Counter class in Python, including lists, tuples, strings, and dictionaries (which count the keys)

The Counter constructor iterates through the input, using each unique element as a dictionary key and storing its frequency as the value. If the input is a dictionary, the keys become the elements and the values become the initial counts.

Lists and Tuples: Pass a collection of items to count their frequencies (e.g., Counter([1, 2, 1, 3])). 

Strings: Treat the string as a sequence of characters to count letter occurrences (e.g., Counter("hello")). 

Dictionaries: Initialize a Counter with a dictionary where keys are elements and values are initial counts (e.g., Counter({'a': 10, 'b': 11})). 

Keyword Arguments: Pass keyword arguments directly to initialize counts (e.g., Counter(a=1, b=2)). 

Another Counter: You can pass an existing Counter object to create a copy or initialize a new one.

A Counter is a specialized dictionary designed for one job: Counting hashable objects.

Key Features of an Elite Engineer’s Toolkit:.most_common(n): Returns the $n$ most frequent elements. This is the "Top-K" problem in computer science, used everywhere from YouTube recommendations to leaderboard logic.

Mathematical Operators: You can add (+), subtract (-), and find the intersection (&) or union (|) of two Counters.

Automatic Initialization: Like defaultdict, it returns 0 for missing keys instead of raising a KeyError.

2. The "YouTube Concept": The Voting Booth
If you were explaining this on YouTube, use the Voting Booth analogy:

Standard Dict: You have to build the ballot box yourself, check if the candidate's name is on the box, and if not, write it on a new folder.

Counter: It’s a smart ballot box. You just drop the name in. At the end of the night, you don't manually count; you just ask the box, "Who are the top 3 winners?" using .most_common(3).

```python
def count_votes(lst):
    count_dict = {}
    for element in lst:
        if element in count_dict:
            count_dict[element] += 1
        else:
            count_dict[element] = 1
    return count_dict

data = ['BJP','BJP','AAP','TMC','BJP','BJP','AAP']
count_votes(data)
```

You cannot pass a 2D or 3D list/tuple directly to Counter to count all nested elements. It will treat each inner list/tuple as a single, unhashable (for lists) or hashable (for tuples) object. 

For 2D/3D Lists: Lists are unhashable, so Counter([[1, 2], [1, 2]]) will raise a TypeError. 
For 2D/3D Tuples: Tuples are hashable, so Counter(((1, 2), (1, 2))) will count the tuples themselves, resulting in Counter({(1, 2): 2}), not the individual numbers.
